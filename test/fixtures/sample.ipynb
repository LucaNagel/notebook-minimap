# Plot linear cryo in vivo kidney pH imaging

## Load data

```python
import numpy as np
import matplotlib.pyplot as plt
import hypermri
from study_config import *

csi = hypermri.load('scan_27')
print(csi.shape)
```

```python
plt.plot(csi.freq_axis_ppm, csi.mean_spectrum())
plt.gca().invert_xaxis()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(12, 4))
for i, a in enumerate(ax):
    a.imshow(anat[i], cmap='gray')
    a.contour(mask[i], colors='r')
```
